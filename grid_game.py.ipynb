```python

import random

# Tahtayı oluşturma
def create_grid(size):
    return [[" " for _ in range(size)] for _ in range(size)]


# Tahtayı yazdırma
def print_grid(grid):
    size = len(grid)
    for i in range(size):
        print(" ".join(grid[i]))
    print()


# Belirli bir pozisyona nesne yerleştirme
def place_object(grid, symbol, x, y):
    grid[x][y] = symbol


# Oyuncudan hedefe bir yol olup olmadığını kontrol etme
def path_exists(grid, x, y, target_x, target_y, visited):
    size = len(grid)

    if x < 0 or x >= size or y < 0 or y >= size:
        return False

    if grid[x][y] == "X" or visited[x][y]:
        return False

    if x == target_x and y == target_y:
        return True

    visited[x][y] = True

    if path_exists(grid, x - 1, y, target_x, target_y, visited):
        return True
    if path_exists(grid, x + 1, y, target_x, target_y, visited):
        return True
    if path_exists(grid, x, y - 1, target_x, target_y, visited):
        return True
    if path_exists(grid, x, y + 1, target_x, target_y, visited):
        return True

    return False


# Rastgele engeller yerleştirme
def place_random_obstacles(grid, num_obstacles):
    size = len(grid)
    obstacles_placed = 0

    while obstacles_placed < num_obstacles:
        x = random.randint(0, size - 1)
        y = random.randint(0, size - 1)

        if grid[x][y] == " ":
            grid[x][y] = "X"

            visited = [
                [False for _ in range(size)]
                for _ in range(size)
            ]

            # Engel yerleştirildikten sonra yol var mı kontrol et
            if path_exists(grid, 0, 0, size - 1, size - 1, visited):
                obstacles_placed += 1
            else:
                # Yol kapanıyorsa engeli kaldır
                grid[x][y] = " "


# Karakteri hareket ettirme
def move_character(grid, character, direction):
    size = len(grid)

    # Karakterin mevcut pozisyonunu bul
    current_x = -1
    current_y = -1

    for i in range(size):
        for j in range(size):
            if grid[i][j] == character:
                current_x = i
                current_y = j
                break
        if current_x != -1:
            break

    # Hareket yönüne göre yeni pozisyon hesapla
    new_x, new_y = current_x, current_y

    if direction == "u":
        new_x -= 1
    elif direction == "d":
        new_x += 1
    elif direction == "l":
        new_y -= 1
    elif direction == "r":
        new_y += 1

    # Sınır kontrolü
    if (new_x < 0 or new_x >= size or
            new_y < 0 or new_y >= size):
        print("Geçersiz hareket! Tahtanın dışına çıkamazsınız.")
        return False

    # Engel kontrolü
    if grid[new_x][new_y] == "X":
        print("Oyun Bitti! Engelle çarpıştınız.")

        # Oyuncuyu tahtadan sil
        grid[current_x][current_y] = " "
        return True

    # Hedef kontrolü
    elif grid[new_x][new_y] == "H":
        print("Hedefe ulaştınız! Tebrikler, oyunu kazandınız.")
        print("Oyun bitti. Teşekkürler!")

        # Oyuncuyu hedef hücresine taşı
        grid[current_x][current_y] = " "
        grid[new_x][new_y] = character
        return True

    # Normal hareket
    grid[current_x][current_y] = " "
    grid[new_x][new_y] = character

    return False


# Oyun döngüsü
def play_game(size, num_obstacles):
    grid = create_grid(size)

    # Oyuncu ve hedef pozisyonları
    player_x, player_y = 0, 0
    target_x, target_y = size - 1, size - 1

    # Oyuncu ve hedefi yerleştir
    place_object(grid, "P", player_x, player_y)
    place_object(grid, "H", target_x, target_y)

    # Rastgele engeller yerleştir
    place_random_obstacles(grid, num_obstacles)

    print("2D Simülasyon Oyununa Hoşgeldiniz!")
    print("Oyunun başında bir karakter seçeceksiniz.")
    print("Hedefe (H) ulaşmaya çalışın, engele (X) çarpmamaya dikkat edin.")
    print("Komutlar: hareket için: h + [u/ d/ r/ l], çık")
    print("Lütfen bir karakter seçin (Ör: P): P")
    print()
    print("Oyun Başladı!")
    print_grid(grid)

    game_over = False

    while not game_over:
        command = input("Komut: ").strip().lower()

        # Oyundan çıkma
        if command == "çık":
            print("Oyun kapatılıyor. Teşekkürler!")
            break

        # Komut formatını kontrol et: h + boşluk + yön
        if (len(command) == 3 and
                command[0] == "h" and
                command[1] == " " and
                command[2] in ["u", "d", "l", "r"]):

            direction = command[2]

            game_over = move_character(grid, "P", direction)

            # Her hareket girişinden sonra tahtayı göster
            print_grid(grid)

        else:
            print("Geçersiz komut! Komutlar: h + [u/ d/ r/ l], çık")


# Oyunu başlat
play_game(5, 3)
```
